```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.agents import create_agent

def whats_up(name:str) -> str :

    """Use this tool STRICTLY ONLY when the user says their name is exactly 'Rakesh'. 
    Do NOT use this tool for any other name or person. If the person is not Rakesh, 
    do not use this tool and just reply normally."""

    return f"{name} is an undergraduate computer science and engineering student specializing in artificial intelligence and machine learning at the Heritage Institute of Technology in Kolkata for the 2024–2028 batch, who actively builds and learns backend systems, participates in collaborative programming and hackathons like EY Techathon 5.0 focusing on financial inclusion AI solutions, maintains an active developer profile on GitHub, and connects professionally via LinkedIn, representing an emerging tech talent in the Greater Kolkata Area whose concise profile encapsulates a dedication to modern software engineering, artificial intelligence model applications, and continuous academic growth within the regional engineering ecosystem."

agent = create_agent(
    model="google_genai:gemini-3.6-flash",
    tools=[whats_up],
    system_prompt="you my roomate santanau."
)

result = agent.invoke(
    {"messages": [{"role": "user", "content": "Hi i am Rakesh"}]}
)
result["messages"][-1].content[0]["text"]
```
