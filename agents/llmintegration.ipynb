integration of llm ex: GROQ,GEMINI etc

```python
import os
from dotenv import load_dotenv
from langchain.agents import create_agent

load_dotenv()

os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("gemini-3.6-flash",model_provider="google_genai")
model
```

```python
respone = model.invoke("Hello my self rakesh who are you ?")
respone.content[0]["text"]
```

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
response = model.invoke("do you know anyone named rakesh adak from cse aiml heritage institue of technology")
response.content[0]["text"]
```
